```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate

# 모델을 생성합니다.
llm = ChatOpenAI(model_name="gpt-3.5-turbo")

# 프롬프트를 생성합니다.
prompt = PromptTemplate.from_template("{flower} 에 대해서 200자 내외로 요약해줘")

# 체인을 생성합니다.
chain = prompt | llm
```

```python
%%time 
response = chain.invoke({"flower": "장미"})
print(response.content)
```

```python
#langchain.prompts=>langchain_classic.prompts
import os
from langchain_openai import ChatOpenAI
from langchain_classic.prompts import PromptTemplate

# 프롬프트 템플릿을 사용하여 질문을 생성합니다.
prompt = PromptTemplate.from_template("{fruit}의 색상이 무엇입니까?")
```

```python
# class에 대해 직렬화가 가능한지 체크합니다.
print(f"ChatOpenAI: {ChatOpenAI.is_lc_serializable()}")
```

```python
llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)

# llm에 대해 직렬화가 가능한지 체크합니다.
print(f"ChatOpenAI: {llm.is_lc_serializable()}")
```

```python
# 체인을 생성합니다.
chain = prompt | llm

# 직렬화가 가능한지 체크합니다.
chain.is_lc_serializable()
```

```python
#체인 직렬화하기(객체를 딕셔너리나 Json 문자열로 변화하는 과정/ dumpd-딕셔너리 dumps-str문자열)
from langchain_core.load import dumpd, dumps

dumpd_chain = dumpd(chain)
dumpd_chain
```

```python
type(dumpd_chain)
```

```python
dumps_chain = dumps(chain)
dumps_chain
```

```python
type(dumps_chain)
```

```python
#pickle 파일로 직렬화하구 로드하기
#직렬화 = Python 객체를 저장/전송 가능한 데이터로 바꾸는 것
#역직렬화 = 그 데이터를 다시 Python 객체로 복원하는 것
#pickle = Python 객체를 바이너리(0/1) 형태로 직렬화/역직렬화할 때 사용하는 대표적인 포맷
```

```python
#wb: 바이너리 쓰기모드 f: 파일 객체
import pickle

with open("fruit_chain.pkl", "wb") as f:
    pickle.dump(dumpd_chain, f)
```

```python
#w: 쓰기모드 fp: 파일 객체 
import json

with open("fruit_chain.json", "w") as fp:
    json.dump(dumpd_chain, fp)
```

```python
#rb: 바이너리 읽기 모드
import pickle

# pickle 파일을 로드
with open("fruit_chain.pkl", "rb") as f:
    loaded_chain = pickle.load(f)
```

```python
from langchain_core.load import load

# 체인을 로드
chain_from_file = load(loaded_chain, allowed_objects='all')

# 체인을 실행
print(chain_from_file.invoke({"fruit": "사과"}))
```

```python
from langchain_core.load import load, loads

load_chain = load(
    loaded_chain, allowed_objects='all', secrets_map={"OPENAI_API_KEY": os.environ["OPENAI_API_KEY"]}
)

# 불러온 체인이 정상 동작하는지 확인합니다.
load_chain.invoke({"fruit": "사과"})
```

```python
with open("fruit_chain.json", "r") as fp:
    loaded_from_json_chain = json.load(fp)
    loads_chain = load(loaded_from_json_chain, allowed_objects='all')
```

```python
# 불러온 체인이 정상 동작하는지 확인합니다.
loads_chain.invoke({"fruit": "사과"})
```
